# SegNet — FINAL 20-PATIENT LOCKED-TEST ENSEMBLE

This notebook performs the **final untouched 20-patient locked-test evaluation** for the reimplemented SegNet baseline.

## Frozen protocol — do not modify after seeing test results

Primary inference rule:

1. load the five frozen Fold 0–4 `best_model.pt` checkpoints;
2. run all five models on each locked-test patient;
3. average their 4-class softmax probabilities with equal weight;
4. use `argmax` of the averaged probability map;
5. **no threshold tuning**;
6. **no calibration fitting on the test set**;
7. **no extra post-processing or test-set optimization**.

The inference rule was fixed before this baseline's locked-test evaluation.

```python
@lru_cache(maxsize=48)
def load_volume(path):
    return np.asarray(
        nib.load(
            path,
            mmap=True,
        ).dataobj,
        dtype=np.float32,
    )


@lru_cache(maxsize=36)
def load_normalized_volume(path):
    array = load_volume(path).copy()

    finite = np.isfinite(array)
    nonzero = (
        finite
        & (
            np.abs(array)
            > 1e-6
        )
    )

    if nonzero.sum() < 100:
        return np.nan_to_num(
            array
        ).astype(np.float32)

    low, high = np.percentile(
        array[nonzero],
        [
            0.5,
            99.5,
        ],
    )

    high = max(
        high,
        low + 1.0,
    )

    array = np.clip(
        array,
        low,
        high,
    )

    array = (
        array
        - array[nonzero].mean()
    ) / (
        array[nonzero].std()
        + 1e-6
    )

    array[~finite] = 0.0

    return array.astype(
        np.float32
    )


@lru_cache(maxsize=48)
def load_binary_mask(path):
    return (
        load_volume(path)
        > 0
    ).astype(np.uint8)


def compose_target(row):
    ventricle = load_binary_mask(
        row.vent
    )

    normal_wmh = load_binary_mask(
        row.nwmh
    )

    abnormal_wmh = load_binary_mask(
        row.abwmh
    )

    target = np.zeros(
        ventricle.shape,
        dtype=np.uint8,
    )

    target[
        ventricle
        > 0
    ] = 1

    target[
        normal_wmh
        > 0
    ] = 2

    target[
        abnormal_wmh
        > 0
    ] = 3

    return target


def triplet(
    volume,
    slice_index,
):
    depth = volume.shape[2]

    indices = [
        max(
            0,
            slice_index - 1,
        ),
        slice_index,
        min(
            depth - 1,
            slice_index + 1,
        ),
    ]

    return np.stack(
        [
            volume[
                :,
                :,
                index,
            ]
            for index in indices
        ],
        axis=0,
    )


def build_patient_inputs(row):
    flair = load_normalized_volume(
        row.flair
    )

    t1 = load_normalized_volume(
        row.t1
    )

    t2 = load_normalized_volume(
        row.t2
    )

    depth = flair.shape[2]

    inputs = []

    for slice_index in range(depth):
        inputs.append(
            np.concatenate(
                [
                    triplet(
                        flair,
                        slice_index,
                    ),
                    triplet(
                        t1,
                        slice_index,
                    ),
                    triplet(
                        t2,
                        slice_index,
                    ),
                ],
                axis=0,
            )
        )

    return np.stack(
        inputs,
        axis=0,
    ).astype(np.float32)
```

```python

class BaselineConvBlock(nn.Module):
    def __init__(self, in_ch, out_ch, repeats=2):
        super().__init__()
        layers = []
        current = in_ch

        for _ in range(repeats):
            groups = 8
            while out_ch % groups != 0 and groups > 1:
                groups //= 2

            layers.extend([
                nn.Conv2d(
                    current,
                    out_ch,
                    kernel_size=3,
                    padding=1,
                    bias=False,
                ),
                nn.GroupNorm(groups, out_ch),
                nn.SiLU(inplace=True),
            ])
            current = out_ch

        self.net = nn.Sequential(*layers)

    def forward(self, x):
        return self.net(x)


class BaselineOutputMixin:
    @staticmethod
    def pack(logits):
        return {
            "segmentation_logits": logits,
            "aux_logits_d2": None,
            "aux_logits_d1": None,
        }

class SegNetEncoderStage(nn.Module):
    def __init__(self, in_ch, out_ch, repeats):
        super().__init__()
        self.block = BaselineConvBlock(
            in_ch,
            out_ch,
            repeats=repeats,
        )
        self.pool = nn.MaxPool2d(
            kernel_size=2,
            stride=2,
            return_indices=True,
        )

    def forward(self, x):
        x = self.block(x)
        size = x.size()
        pooled, indices = self.pool(x)
        return pooled, indices, size


class SegNetDecoderStage(nn.Module):
    def __init__(self, in_ch, out_ch, repeats):
        super().__init__()
        self.unpool = nn.MaxUnpool2d(
            kernel_size=2,
            stride=2,
        )
        self.block = BaselineConvBlock(
            in_ch,
            out_ch,
            repeats=repeats,
        )

    def forward(self, x, indices, output_size):
        x = self.unpool(
            x,
            indices,
            output_size=output_size,
        )
        return self.block(x)


class SegNetBaseline(nn.Module):
    """
    SegNet-style 2.5D multimodal baseline.

    Encoder pooling indices are reused by the decoder for non-learned
    spatial upsampling. No skip-feature concatenation is used.
    """

    def __init__(self, base=24, num_classes=4, **kwargs):
        super().__init__()

        b = int(base)
        c1, c2, c3, c4, c5 = (
            b,
            b * 2,
            b * 4,
            b * 8,
            b * 12,
        )

        self.enc1 = SegNetEncoderStage(9, c1, repeats=2)
        self.enc2 = SegNetEncoderStage(c1, c2, repeats=2)
        self.enc3 = SegNetEncoderStage(c2, c3, repeats=3)
        self.enc4 = SegNetEncoderStage(c3, c4, repeats=3)
        self.enc5 = SegNetEncoderStage(c4, c5, repeats=3)

        self.dec5 = SegNetDecoderStage(c5, c4, repeats=3)
        self.dec4 = SegNetDecoderStage(c4, c3, repeats=3)
        self.dec3 = SegNetDecoderStage(c3, c2, repeats=3)
        self.dec2 = SegNetDecoderStage(c2, c1, repeats=2)
        self.dec1 = SegNetDecoderStage(c1, c1, repeats=2)

        self.head = nn.Conv2d(
            c1,
            num_classes,
            kernel_size=1,
        )

    def forward(self, x):
        x, i1, s1 = self.enc1(x)
        x, i2, s2 = self.enc2(x)
        x, i3, s3 = self.enc3(x)
        x, i4, s4 = self.enc4(x)
        x, i5, s5 = self.enc5(x)

        x = self.dec5(x, i5, s5)
        x = self.dec4(x, i4, s4)
        x = self.dec3(x, i3, s3)
        x = self.dec2(x, i2, s2)
        x = self.dec1(x, i1, s1)

        return BaselineOutputMixin.pack(
            self.head(x)
        )


# Compatibility alias used by the shared training loop.
VentiMorphRelNet = SegNetBaseline
```

```python
def build_model_from_checkpoint(checkpoint_path, expected_fold):
    checkpoint = torch.load(
        checkpoint_path,
        map_location="cpu",
        weights_only=False,
    )

    if "model_state" not in checkpoint:
        raise KeyError(f"{checkpoint_path} does not contain model_state.")

    config = checkpoint.get("config", {})
    if int(config.get("fold", expected_fold)) != expected_fold:
        raise RuntimeError(
            f"Checkpoint/config fold mismatch: expected {expected_fold}, "
            f"found {config.get('fold')}."
        )

    model = VentiMorphRelNet(
        base=int(config.get("base", 24)),
    ).to(DEVICE)

    incompatible = model.load_state_dict(
        checkpoint["model_state"],
        strict=True,
    )

    nonfinite = [
        name for name, tensor in checkpoint["model_state"].items()
        if torch.is_tensor(tensor)
        and not torch.isfinite(tensor).all()
    ]
    if nonfinite:
        raise RuntimeError(
            f"Fold {expected_fold} checkpoint contains non-finite tensors: "
            f"{nonfinite[:10]}"
        )

    model.eval()
    for parameter in model.parameters():
        parameter.requires_grad_(False)

    return model, checkpoint, config, incompatible


@torch.inference_mode()
def predict_patient(model, patient_id):
    patient_id = str(patient_id).zfill(3)
    row = manifest_by_id.loc[patient_id]
    inputs = build_patient_inputs(row)

    prediction_batches = []

    for start in range(0, len(inputs), INFERENCE_BATCH_SIZE):
        batch = torch.from_numpy(
            inputs[start:start + INFERENCE_BATCH_SIZE]
        ).to(DEVICE)

        with torch.autocast(
            device_type=DEVICE.type,
            enabled=(DEVICE.type == "cuda"),
        ):
            output = model(batch)
            prediction = torch.argmax(
                output["segmentation_logits"],
                dim=1,
            )

        prediction_batches.append(
            prediction.to(torch.uint8).cpu().numpy()
        )

        del batch, output, prediction

    prediction = np.transpose(
        np.concatenate(prediction_batches, axis=0),
        (1, 2, 0),
    ).astype(np.uint8)

    target = compose_target(row)

    if prediction.shape != target.shape:
        raise RuntimeError(
            f"Prediction/target shape mismatch for patient {patient_id}: "
            f"{prediction.shape} vs {target.shape}"
        )

    return prediction, target
```

```python
def safe_divide(
    numerator,
    denominator,
    empty_value=0.0,
):
    if denominator == 0:
        return float(
            empty_value
        )

    return float(
        numerator
        / denominator
    )


def binary_metrics(
    prediction,
    target,
):
    prediction = prediction.astype(
        bool
    )

    target = target.astype(
        bool
    )

    true_positive = int(
        (
            prediction
            & target
        ).sum()
    )

    false_positive = int(
        (
            prediction
            & ~target
        ).sum()
    )

    false_negative = int(
        (
            ~prediction
            & target
        ).sum()
    )

    true_negative = int(
        (
            ~prediction
            & ~target
        ).sum()
    )

    if (
        prediction.sum() == 0
        and target.sum() == 0
    ):
        dice = 1.0
        iou = 1.0
        precision = 1.0
        recall = 1.0

    else:
        dice = safe_divide(
            2
            * true_positive,
            2
            * true_positive
            + false_positive
            + false_negative,
        )

        iou = safe_divide(
            true_positive,
            true_positive
            + false_positive
            + false_negative,
        )

        precision = safe_divide(
            true_positive,
            true_positive
            + false_positive,
        )

        recall = safe_divide(
            true_positive,
            true_positive
            + false_negative,
        )

    specificity = safe_divide(
        true_negative,
        true_negative
        + false_positive,
        empty_value=1.0,
    )

    return {
        "dice": dice,
        "iou": iou,
        "precision": precision,
        "recall": recall,
        "specificity": specificity,
        "tp": true_positive,
        "fp": false_positive,
        "fn": false_negative,
        "tn": true_negative,
    }


def surface_distances(
    prediction,
    target,
    spacing,
):
    prediction = prediction.astype(
        bool
    )

    target = target.astype(
        bool
    )

    if (
        not prediction.any()
        and not target.any()
    ):
        return np.array(
            [
                0.0,
            ],
            dtype=np.float64,
        )

    if (
        not prediction.any()
        or not target.any()
    ):
        return np.array(
            [
                np.nan,
            ],
            dtype=np.float64,
        )

    structure = ndimage.generate_binary_structure(
        3,
        1,
    )

    prediction_surface = (
        prediction
        ^ ndimage.binary_erosion(
            prediction,
            structure=structure,
            border_value=0,
        )
    )

    target_surface = (
        target
        ^ ndimage.binary_erosion(
            target,
            structure=structure,
            border_value=0,
        )
    )

    target_distance = ndimage.distance_transform_edt(
        ~target_surface,
        sampling=spacing,
    )

    prediction_distance = ndimage.distance_transform_edt(
        ~prediction_surface,
        sampling=spacing,
    )

    distances = np.concatenate([
        target_distance[
            prediction_surface
        ],
        prediction_distance[
            target_surface
        ],
    ])

    return distances.astype(
        np.float64
    )


def distance_metrics(
    prediction,
    target,
    spacing,
):
    distances = surface_distances(
        prediction,
        target,
        spacing,
    )

    if np.isnan(
        distances
    ).all():
        return {
            "hd95_mm": np.nan,
            "assd_mm": np.nan,
        }

    return {
        "hd95_mm": float(
            np.nanpercentile(
                distances,
                95,
            )
        ),
        "assd_mm": float(
            np.nanmean(
                distances
            )
        ),
    }


def connected_components(
    mask,
    minimum_voxels=3,
):
    structure = ndimage.generate_binary_structure(
        3,
        2,
    )

    labels, number = ndimage.label(
        mask.astype(bool),
        structure=structure,
    )

    components = []

    for component_id in range(
        1,
        number + 1,
    ):
        component = (
            labels
            == component_id
        )

        size = int(
            component.sum()
        )

        if size >= minimum_voxels:
            components.append({
                "mask": component,
                "voxels": size,
            })

    return components


def lesion_matching_metrics(
    prediction,
    target,
    minimum_voxels=3,
    iou_threshold=0.10,
):
    predicted_components = connected_components(
        prediction,
        minimum_voxels,
    )

    target_components = connected_components(
        target,
        minimum_voxels,
    )

    number_predicted = len(
        predicted_components
    )

    number_target = len(
        target_components
    )

    if (
        number_predicted == 0
        and number_target == 0
    ):
        return {
            "lesion_sensitivity": 1.0,
            "lesion_precision": 1.0,
            "lesion_f1": 1.0,
            "true_lesions": 0,
            "predicted_lesions": 0,
            "matched_lesions": 0,
            "false_positive_lesions": 0,
            "small_lesion_recall": np.nan,
            "medium_lesion_recall": np.nan,
            "large_lesion_recall": np.nan,
        }

    if (
        number_predicted == 0
        or number_target == 0
    ):
        matched_target_indices = set()

    else:
        iou_matrix = np.zeros(
            (
                number_target,
                number_predicted,
            ),
            dtype=np.float64,
        )

        for target_index, target_component in enumerate(
            target_components
        ):
            target_mask = target_component[
                "mask"
            ]

            for predicted_index, predicted_component in enumerate(
                predicted_components
            ):
                predicted_mask = predicted_component[
                    "mask"
                ]

                intersection = int(
                    (
                        target_mask
                        & predicted_mask
                    ).sum()
                )

                union = int(
                    (
                        target_mask
                        | predicted_mask
                    ).sum()
                )

                if union > 0:
                    iou_matrix[
                        target_index,
                        predicted_index,
                    ] = (
                        intersection
                        / union
                    )

        target_indices, predicted_indices = linear_sum_assignment(
            -iou_matrix
        )

        valid_matches = [
            (
                target_index,
                predicted_index,
            )
            for target_index, predicted_index
            in zip(
                target_indices,
                predicted_indices,
            )
            if iou_matrix[
                target_index,
                predicted_index,
            ]
            >= iou_threshold
        ]

        matched_target_indices = {
            target_index
            for target_index, _
            in valid_matches
        }

    matched_lesions = len(
        matched_target_indices
    )

    sensitivity = safe_divide(
        matched_lesions,
        number_target,
        empty_value=1.0,
    )

    precision = safe_divide(
        matched_lesions,
        number_predicted,
        empty_value=(
            1.0
            if number_target == 0
            else 0.0
        ),
    )

    lesion_f1 = safe_divide(
        2
        * precision
        * sensitivity,
        precision
        + sensitivity,
    )

    size_groups = {
        "small": [],
        "medium": [],
        "large": [],
    }

    for target_index, component in enumerate(
        target_components
    ):
        voxels = component[
            "voxels"
        ]

        if voxels <= 20:
            group = "small"

        elif voxels <= 100:
            group = "medium"

        else:
            group = "large"

        size_groups[
            group
        ].append(
            target_index
        )

    size_recall = {}

    for group, indices in size_groups.items():
        if not indices:
            size_recall[
                group
            ] = np.nan

        else:
            detected = sum(
                index
                in matched_target_indices
                for index
                in indices
            )

            size_recall[
                group
            ] = (
                detected
                / len(indices)
            )

    return {
        "lesion_sensitivity": sensitivity,
        "lesion_precision": precision,
        "lesion_f1": lesion_f1,
        "true_lesions": number_target,
        "predicted_lesions": number_predicted,
        "matched_lesions": matched_lesions,
        "false_positive_lesions": max(
            0,
            number_predicted
            - matched_lesions,
        ),
        "small_lesion_recall": size_recall[
            "small"
        ],
        "medium_lesion_recall": size_recall[
            "medium"
        ],
        "large_lesion_recall": size_recall[
            "large"
        ],
    }


def periventricular_fp_rate(
    predicted_abnormal,
    true_abnormal,
    true_ventricle,
    spacing_xy,
    maximum_distance_mm=10.0,
):
    distance = np.zeros_like(
        true_ventricle,
        dtype=np.float32,
    )

    for slice_index in range(
        true_ventricle.shape[2]
    ):
        distance[
            :,
            :,
            slice_index,
        ] = ndimage.distance_transform_edt(
            ~true_ventricle[
                :,
                :,
                slice_index,
            ].astype(bool),
            sampling=spacing_xy,
        )

    periventricular_region = (
        distance
        <= maximum_distance_mm
    ) & (
        ~true_ventricle.astype(bool)
    )

    periventricular_predictions = (
        predicted_abnormal
        & periventricular_region
    )

    false_positive = (
        predicted_abnormal
        & ~true_abnormal
        & periventricular_region
    )

    return safe_divide(
        int(
            false_positive.sum()
        ),
        int(
            periventricular_predictions.sum()
        ),
        empty_value=0.0,
    )


def update_confusion_matrix(
    confusion_matrix,
    target,
    prediction,
):
    encoded = (
        target.astype(np.int64)
        * 4
        + prediction.astype(np.int64)
    )

    counts = np.bincount(
        encoded.ravel(),
        minlength=16,
    ).reshape(
        4,
        4,
    )

    confusion_matrix += counts
```

```python
# Load all five frozen best checkpoints.
ensemble_models = []
checkpoint_audit_rows = []

for fold in range(5):
    model, checkpoint, config, incompatible = build_model_from_checkpoint(
        CHECKPOINT_BUNDLES[fold]["checkpoint"],
        expected_fold=fold,
    )

    ensemble_models.append(model)

    checkpoint_audit_rows.append({
        "fold": fold,
        "checkpoint": str(CHECKPOINT_BUNDLES[fold]["checkpoint"]),
        "checkpoint_epoch": checkpoint.get("epoch"),
        "checkpoint_best_epoch": checkpoint.get("best_epoch"),
        "parameters": int(sum(p.numel() for p in model.parameters())),
        "base": config.get("base"),
        "strict_load_status": (
            "strict=True load completed"
            if incompatible is None
            else str(incompatible)
        ),
    })

checkpoint_audit_df = pd.DataFrame(checkpoint_audit_rows)

if len(ensemble_models) != 5:
    raise RuntimeError("Exactly five frozen fold models are required.")

if checkpoint_audit_df["fold"].tolist() != [0, 1, 2, 3, 4]:
    raise RuntimeError("Fold checkpoint order is not exactly 0,1,2,3,4.")

checkpoint_audit_df.to_csv(
    TABLE_DIR / "segnet_locked_test_checkpoint_audit.csv",
    index=False,
)

print("Loaded five frozen SegNet fold checkpoints.")
print("Final inference: equal-weight softmax ensemble -> argmax.")
display(checkpoint_audit_df)
```

```python
@torch.inference_mode()
def predict_patient_ensemble(patient_id):
    patient_id = str(patient_id).zfill(3)
    row = manifest_by_id.loc[patient_id]

    inputs = build_patient_inputs(row)
    target = compose_target(row)

    prediction_batches = []

    for start in range(0, len(inputs), INFERENCE_BATCH_SIZE):
        batch = torch.from_numpy(
            inputs[start:start + INFERENCE_BATCH_SIZE]
        ).to(DEVICE)

        probability_sum = None

        for model in ensemble_models:
            with torch.autocast(
                device_type=DEVICE.type,
                enabled=(DEVICE.type == "cuda"),
            ):
                output = model(batch)
                probabilities = torch.softmax(
                    output["segmentation_logits"].float(),
                    dim=1,
                )

            if probability_sum is None:
                probability_sum = probabilities
            else:
                probability_sum = probability_sum + probabilities

            del output, probabilities

        mean_probability = probability_sum / float(len(ensemble_models))
        prediction = torch.argmax(
            mean_probability,
            dim=1,
        )

        prediction_batches.append(
            prediction.to(torch.uint8).cpu().numpy()
        )

        del batch, probability_sum, mean_probability, prediction

    prediction = np.transpose(
        np.concatenate(
            prediction_batches,
            axis=0,
        ),
        (1, 2, 0),
    ).astype(np.uint8)

    if prediction.shape != target.shape:
        raise RuntimeError(
            f"Prediction/target shape mismatch for patient {patient_id}: "
            f"{prediction.shape} vs {target.shape}"
        )

    return prediction, target
```

```python
CLASS_INFORMATION = {
    1: "ventricle",
    2: "normal_wmh",
    3: "abnormal_wmh",
}

patient_rows = []
lesion_rows = []
global_confusion = np.zeros((4, 4), dtype=np.int64)

print("\nFINAL LOCKED-TEST INFERENCE STARTED")
print("Primary inference rule: equal-weight five-model softmax ensemble -> argmax")
print("No threshold tuning or calibration fitting is performed.")

for patient_number, patient_id in enumerate(
    locked_test_ids,
    start=1,
):
    prediction, target = predict_patient_ensemble(
        patient_id
    )

    row = manifest_by_id.loc[patient_id]
    reference_image = nib.load(
        row.flair,
        mmap=True,
    )
    spacing = tuple(
        float(v)
        for v in reference_image.header.get_zooms()[:3]
    )

    update_confusion_matrix(
        global_confusion,
        target,
        prediction,
    )

    patient_metrics = {
        "patient_id": patient_id,
        "evaluation_set": "locked_test",
        "ensemble_models": 5,
    }

    for class_id, class_name in CLASS_INFORMATION.items():
        pred_binary = prediction == class_id
        true_binary = target == class_id

        overlap = binary_metrics(
            pred_binary,
            true_binary,
        )
        distance = distance_metrics(
            pred_binary,
            true_binary,
            spacing=spacing,
        )

        for metric_name in [
            "dice",
            "iou",
            "precision",
            "recall",
            "specificity",
            "tp",
            "fp",
            "fn",
            "tn",
        ]:
            patient_metrics[
                f"{class_name}_{metric_name}"
            ] = overlap[metric_name]

        patient_metrics[
            f"{class_name}_hd95_mm"
        ] = distance["hd95_mm"]
        patient_metrics[
            f"{class_name}_assd_mm"
        ] = distance["assd_mm"]

        if class_id in (2, 3):
            lesion = lesion_matching_metrics(
                pred_binary,
                true_binary,
                minimum_voxels=3,
                iou_threshold=0.10,
            )

            for metric_name, metric_value in lesion.items():
                patient_metrics[
                    f"{class_name}_{metric_name}"
                ] = metric_value

            lesion_rows.append({
                "patient_id": patient_id,
                "class": class_name,
                **lesion,
            })

    patient_metrics["mean_foreground_dice"] = float(
        np.mean([
            patient_metrics["ventricle_dice"],
            patient_metrics["normal_wmh_dice"],
            patient_metrics["abnormal_wmh_dice"],
        ])
    )

    true_ventricle = load_binary_mask(
        row.vent
    ).astype(bool)

    patient_metrics[
        "periventricular_abnormal_fp_rate"
    ] = periventricular_fp_rate(
        prediction == 3,
        target == 3,
        true_ventricle,
        spacing[:2],
    )

    patient_rows.append(patient_metrics)

    # Save the final ensemble segmentation with the FLAIR geometry.
    prediction_image = nib.Nifti1Image(
        prediction.astype(np.uint8),
        affine=reference_image.affine,
        header=reference_image.header.copy(),
    )
    prediction_image.set_data_dtype(np.uint8)

    nib.save(
        prediction_image,
        PREDICTION_DIR
        / f"patient_{patient_id}_v281_ensemble_prediction.nii.gz",
    )

    print(
        f"[{patient_number:02d}/20] patient {patient_id} | "
        f"FG={patient_metrics['mean_foreground_dice']:.4f} "
        f"V={patient_metrics['ventricle_dice']:.4f} "
        f"N={patient_metrics['normal_wmh_dice']:.4f} "
        f"A={patient_metrics['abnormal_wmh_dice']:.4f}"
    )

    del prediction, target
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

patient_metrics_df = pd.DataFrame(
    patient_rows
)
lesion_metrics_df = pd.DataFrame(
    lesion_rows
)

if len(patient_metrics_df) != 20:
    raise RuntimeError(
        f"Expected 20 locked-test rows, found {len(patient_metrics_df)}."
    )

if patient_metrics_df["patient_id"].nunique() != 20:
    raise RuntimeError(
        "Locked-test output does not contain 20 unique patients."
    )

if set(patient_metrics_df["patient_id"]) != set(locked_test_ids):
    raise RuntimeError(
        "Evaluated patient IDs differ from the frozen locked-test IDs."
    )

patient_metrics_df.to_csv(
    TABLE_DIR / "v281_locked_test_patient_metrics.csv",
    index=False,
)
lesion_metrics_df.to_csv(
    TABLE_DIR / "v281_locked_test_lesion_metrics.csv",
    index=False,
)

print("\nLocked-test inference complete.")
print("Patients evaluated:", len(patient_metrics_df))
```
